## Esercizio 1 — Durata di una sessione di laboratorio
Scrivere nella cella dell'esercizio 1 il programma che legge in ingresso la durata di una sessione di laboratorio espressa in secondi e la stampa nel formato ore, minuti e secondi, usando soltanto la divisione intera e il resto. Il programma stampa anche la stessa durata espressa in ore decimali con due cifre dopo la virgola, e segnala con un messaggio il caso in cui la durata superi le quattro ore.

Input di esempio: `9296`  
Output atteso:
```bash
02:34:56  
in ore decimali: 2.58
```
Con l'ingresso `15000` va stampata anche la riga `attenzione: la sessione supera le quattro ore`  

**Tempo stimato: 20 minuti**

````python
secondiTot = 9296
ore = secondiTot // 3600
minuti = (secondiTot % 3600) // 60
secondi = secondiTot % 60

print(f"{ore:02d}:{minuti:02d}:{secondi:02d}")

decimale = (secondiTot / 3600)

print(round(decimale, 2))
if secondiTot > 14400: print("attenzione: la sessione supera le 4 ore")
````

## Esercizio 2 — Media dei voti e giudizio
Scrivere nella cella dell'esercizio 2 il programma che legge in ingresso il nome di uno studente, tre voti in virgola mobile e le ore di assenza, calcola la media dei tre voti e stampa il giudizio secondo questa regola: se le ore di assenza superano 50 il giudizio è non valutabile , indipendentemente dalla media; altrimenti il giudizio è ottimo da 8 in su, sufficiente da 6 in
su, insufficiente sotto il 6. La media va stampata con due decimali dentro una f-string, insieme al nome dello studente.

Input di esempio: `Chiara , 7.5 , 8 , 6.5 , 12`  

Output atteso:  
```bash
Chiara: media 7.33, giudizio sufficiente
```

**Tempo stimato: 25 minuti**

````python
nome = 'Chiara'
voti = 7.5, 8, 6.5
oreAss = 12
media = round(sum(voti)/len(voti), 2)
if oreAss > 50: print(f"{nome}: media {media}, giudizio non valutabile")
elif media >= 8: print(f"{nome}: media {media}, giudizio ottimo")
elif media >= 6: print(f"{nome}: media {media}, giudizio sufficiente")
else: print(f"{nome}: media {media}, giudizio insufficiente")
````

## Esercizio 3 — Tavola di conversione incolonnata
Scrivere nella cella dell'esercizio 3 il programma che legge in ingresso un valore iniziale, un valore finale e un passo, tutti interi, e stampa la tavola di conversione da chilometri a miglia e a metri per tutti i valori dell'intervallo, estremi compresi. **Un chilometro corrisponde a 0,621371 miglia**. La tabella ha un'intestazione e una riga di trattini, le colonne sono allineate a destra e i valori in miglia hanno tre decimali. Se il passo non è positivo il programma stampa un messaggio e non produce la tabella.

Input di esempio: `5 , 20 , 5`

Output atteso:
```bash
km miglia metri
------------------------------------
5 3.107 5000
10 6.214 10000
15 9.321 15000
20 12.427 20000
```

**Tempo stimato: 30 minuti**

````python
inizio = 5
fine = 20
passo = 5
valore = inizio

print(f"{'km':>7}{'miglia':>10}{'metri':>10}")
print('-' * 27)
if passo > 0:
    for valore in range(inizio, fine+1, passo):
        print(f"{valore:>7}{round(valore*0.621371, 3):>10}{(valore*1000):>10}")
        valore += passo
else:
    print('Passo non positivo')
````

## Esercizio 4 — Lettura di un codice di postazione
Scrivere nella cella dell'esercizio 4 il programma che legge in ingresso un codice di postazione di laboratorio nel formato LAB2-PC17-2026, dove le prime quattro posizioni indicano il laboratorio, le quattro successive al trattino indicano il calcolatore e le ultime quattro l'anno di installazione. Il programma ripulisce l'ingresso dagli spazi agli estremi, lo porta in maiuscolo ed estrae con lo slicing le tre parti, stampandole su tre righe distinte. Stampa infine la lunghezza del codice e il suo ultimo carattere, ricavato con un indice negativo.

Input di esempio: `lab2-pc17-2026`

Output atteso:
```bash
laboratorio: LAB2
calcolatore: PC17
anno: 2026
lunghezza: 14, ultimo carattere: 6
```
**Tempo stimato: 25 minuti**

````python
postazione = "lab2-pc17-2026"
info = postazione.split('-')
laboratorio = info[0].upper()
calcolatore = info[1].upper()
anno = info[2]

print(f"laboratorio: {laboratorio}")
print(f"calcolatore: {calcolatore}")
print(f"anno: {anno}")
print(f"lunghezza: {len(postazione)}, ultimo carattere: {postazione[len(postazione)-1]}")
````

## Esercizio 5 — Conteggio dei caratteri di una frase
Scrivere nella cella dell'esercizio 5 il programma che legge in ingresso una frase e conta, con un unico ciclo for sui caratteri, quante vocali, quante consonanti, quante cifre e quanti altri caratteri contiene. Il programma stampa i quattro conteggi e la percentuale di vocali sul totale dei caratteri, con un decimale. Le lettere accentate vanno considerate consonanti: è una semplificazione voluta, da dichiarare in un commento. 

Input di esempio: `Il laboratorio 4BI dura 22 ore`

Output atteso:
```bash
vocali: 12, consonanti: 10, cifre: 3, altri: 5
percentuale di vocali: 40.0%
```

**Tempo stimato: 30 minuti**

````python
frase = "Il laboratorio 4BI dura 22 ore"
voc = "aeiou"
tot = 0
nVoc = 0
nCons = 0
nCifre = 0
nAltri = 0

for char in frase.lower():
    if char.isalpha():
        if char in voc: nVoc += 1
        else: nCons += 1
    elif char.isdigit(): nCifre += 1
    else: nAltri += 1
    tot += 1

print(f"vocali: {nVoc}, consonanti: {nCons}, cifre: {nCifre}, altri: {nAltri}")
print(f"percentuale di vocali: {nVoc/tot*100}%")
````
